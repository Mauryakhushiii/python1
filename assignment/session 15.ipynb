```python
#task 1

def get_song_duration_per_minute(total_duration, num_songs):
    try:
        result = total_duration/ num_songs
    except ZeroDivisionError:
        print("Error: Number of songs is zero , connot divide!")
        result = 0
    finally:
        print("Calculation attemt finished.")
    
    return result

print(get_song_duration_per_minute(200,10))
print(get_song_duration_per_minute(200,0))
    
```

```python
#task 2

def price_per_item_calculator():
    try:
        total = float(input("Enter total cart amount: "))
        count = int(input("Enter item count: "))
        price = total / count
        print(f"Price per item: Rs. {price:.2f}")
    except ZeroDivisionError:
        print("Oops ! Item count is 0, cannot divide. Please add items to cart.")
    except ValueError:
        print("Please enter valid numbers.")

price_per_item_calculator()
                      
```

```python
# task 3

class NoOffersApplied(Exception):
    pass

def paytm_cashback_calculator():
    try:
        total_spend = float(input("Enter total spend: "))
        num_offers = int(input("Enter number of offers applied: "))
        
        if num_offers == 0:
            raise NoOffersApplied("No offers were applied, cannot calculate average cashback!")
        
        avg_cashback = total_spend / num_offers
        print(f"Average cashback per offer: Rs. {avg_cashback:.2f}")
    
    except NoOffersApplied as e:
        print(f"Custom Error: {e}")

paytm_cashback_calculator()
```

```python
# task 4

def calculate_average_rating(total_rating, num_reviews):
    try:
        return total_rating / num_reviews
    except ZeroDivisionError:
        print("Error: Number of reviews cannot be zero!")
        return 0
    finally:
        print("Thank you for using the calculator")

print(calculate_average_rating(500, 0))
print(calculate_average_rating(500, 5))
```

```python
# task 5

def safe_divide_for_zomato(bill_amount, num_people):
    try:
        result = bill_amount / num_people
    except ZeroDivisionError:
        print("Error: Number of people cannot be zero for bill split!")
    except TypeError:
        print("Please provide valid numbers")
    else:
        print(f"Each person pays: Rs. {result:.2f}")
        return result
    finally:
        print("Split calculation done")

safe_divide_for_zomato(1500, 3)
safe_divide_for_zomato(1500, 0)
```
