````python
#Write a recursive function in Python called print_playlist_songs(songs) that takes a list of song names (like a Spotify playlist) and prints each song name one by one using recursion.

def print_playlist_songs(songs):
    if not songs:  
        return

    print(songs[0])
    print_playlist_songs(songs[1:])  


playlist = ["Tere Bin", "Mast Magan", "Falak Tak", "Tum Tak"]
print_playlist_songs(playlist)
````

````python
#Create a recursive function count_unread_messages(messages) that takes a nested dictionary representing WhatsApp chat groups and subgroups, and returns the total number of unread messages across all groups.<br><br><em><strong>Hint:</strong> Each group can have a 'count' key for unread messages and a 'subgroups' key with a list of more groups.</em>

def count_unread_messages(messages):
    total = messages.get('count', 0)

    for group in messages.get('subgroups', []):
        total += count_unread_messages(group)

    return total


messages = {
    'count': 5,
    'subgroups': [
        {
            'count': 3,
            'subgroups': [
                {'count': 2},
                {'count': 4}
            ]
        },
        {
            'count': 6,
            'subgroups': [
                {'count': 1}
            ]
        }
    ]
}

print("Total unread messages:", count_unread_messages(messages))
````

````python
#Build a recursive function format_number_short(n) that takes a number (like a follower count on Instagram or YouTube) and returns it as a string in short format: 1500 as '1.5K', 1200000 as '1.2M', 500 as '500'.
def format_number_short(n):
    if abs(n) < 1000:
        return str(n)

    if abs(n) < 1_000_000:
        return format_number_short(n / 1000) + "K"

    if abs(n) < 1_000_000_000:
        return format_number_short(n / 1_000_000) + "M"

    return format_number_short(n / 1_000_000_000) + "B"


print(format_number_short(500))
print(format_number_short(1500))
print(format_number_short(1200000))
print(format_number_short(2500000000))
````
