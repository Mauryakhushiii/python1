```python
songs = ["Tum Hi Ho", "Channa Mereya", "Agar Tum Sath Ho", "Kesariya"]
clean_songs = list(map(lambda songs: songs.lower(), songs))
print(clean_songs)
```

```python
ratings = [4.2, 3.8, 4.5, 2.9, 3.5]
ans = list(filter(lambda ratings: ratings > 4 , ratings))
print(ans)
```

```python
from functools import reduce
prices = [499, 1299, 299, 799]
result = reduce(lambda a,b:a+b, prices)
print(result)
```

```python
followers = [950, 1500, 2500, 1200000]
def format_followers(num):
    if num >= 1000000:
        return str(num / 1000000) + "M"
    elif num >= 1000:
        return str(num / 1000) + "K"
    else:
        return str(num)

result = list(map(format_followers,followers))
print(result)
```
