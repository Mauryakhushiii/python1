```python
#task2
steps = [5000,7500,9200,10500,8000,12000,11000]
i = 0
while i < len(steps):
    if steps[i] > 10000:
        print("First day crossed 10,000 steps:",i + 1)
        break
    i += 1
```

```python
#task3
def team(team_list):
    for team in team_list:
        if len(team) > 6:
            print(team)
team_list = ["mumbai", "indian", "CSK" ,"Rajasthan Royals" ,"Delhi Capitals","RBC"]
team(team_list)
```

```python
#task 5
prices = [500,300,0,700,400,800]
total = 0
for i in prices:
    if i == 0:
        continue
    total = total + i

    if total > 2000:
        break
print("Final total:",total)
```
