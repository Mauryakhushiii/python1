```python
#Download a small sample of your recent Zomato order history as a CSV (or create a mock CSV with restaurant names and order dates), then use pandas' duplicated() function to find and print any duplicate orders based on restaurant name and date.
import pandas as pd

data = {
    'restaurant_name': [
        'Dominos',
        'Pizza Hut',
        'McDonalds',
        'Dominos',
        'KFC',
        'Pizza Hut',
        'Dominos'
    ],
    'order_date': [
        '2026-10-20',
        '2026-10-21',
        '2026-10-22',
        '2026-10-20',
        '2026-10-23',
        '2026-10-21',
        '2026-10-25'
    ]
}

df = pd.DataFrame(data)

df.to_csv('zomato_orders.csv', index=False)
df = pd.read_csv('zomato_orders.csv')

#print(df)
duplicates = df[df.duplicated(
    subset=['restaurant_name', 'order_date'],
    keep=False
)]

print("Duplicate Orders:")
print(duplicates)
         
```

```python
#Given a list of Flipkart product reviews with some duplicate entries, use value_counts() in pandas to identify which review texts are repeated most often and display the top 3 most common duplicate reviews.
import pandas as pd

data = {
    'review_text': [
        'Good product',
        'Excellent quality',
        'Good product',
        'Average product',
        'Good product',
        'Excellent quality',
        'Bad packaging',
        'Excellent quality',
        'Good product',
        'Bad packaging'
    ]
}

df = pd.DataFrame(data)

#print(df)
review_counts = df['review_text'].value_counts()

print(review_counts)
top_3 = review_counts.head(3)

print("Top 3 Most Common Reviews:")
print(top_3)
```

```python
#Create a DataFrame with mock data for Spotify playlists, including playlist names and creator usernames, where some rows are exact duplicates. Use drop_duplicates() to remove duplicate playlists and print the cleaned DataFrame.
import pandas as pd

data = {
    'playlist_name': [
        'Bollywood Hits',
        'English Songs',
        'Romantic Songs',
        'Bollywood Hits',
        'Party Music',
        'English Songs',
        'Lo-fi Music'
    ],
    'creator_username': [
        'khushi01',
        'rahul22',
        'priya10',
        'khushi01',
        'amit45',
        'rahul22',
        'neha12'
    ]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
cleaned_df = df.drop_duplicates()

print("Cleaned DataFrame:")
print(cleaned_df)
```

```python
#Suppose you have a DataFrame of Instagram usernames where some entries have typos (like 'insta_queen', 'insta-queen', 'instaqueen'). Use the replace() function to standardize all these variants to 'instaqueen'.
import pandas as pd

data = {
    'username': [
        'insta_queen',
        'insta-queen',
        'instaqueen',
        'foodie_girl',
        'insta_queen',
        'insta-queen',
        'travel_lover'
    ]
}

df = pd.DataFrame(data)

#print("Original DataFrame:")
#print(df)
df['username'] = df['username'].replace({
    'insta_queen': 'instaqueen',
    'insta-queen': 'instaqueen'
})

print("Cleaned DataFrame:")
print(df)
```

```python
#You have a DataFrame column for payment status from a Paytm-like app with mixed values: 'Yes', 'yes', 'Y', 'No', 'no', 'N', and some with extra spaces. Write code to unify this column so all paid statuses become 1 and all unpaid statuses become 0, trimming whitespace and fixing capitalization where needed.<br><br><em><strong>Hint:</strong> Use str.strip(), str.lower(), and map/replace methods in pandas.</em>

data = {
    'payment_status': [
        'Yes', 'yes', 'Y', 'No', 'no', 'N',
        ' Yes ', ' NO ', '  y  ', ' n '
    ]
}

df = pd.DataFrame(data)

#print(df)
df['payment_status'] = df['payment_status'].str.strip()

df['payment_status'] = df['payment_status'].str.lower()

df['payment_status'] = df['payment_status'].replace({
    'yes': 1,
    'y': 1,
    'no': 0,
    'n': 0
})

print(df)
```
