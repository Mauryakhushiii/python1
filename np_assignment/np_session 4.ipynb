```python
##Download a large Excel file (at least 10,000 rows) of Flipkart product listings. Use pd.read_excel() with the chunksize parameter to read the file in chunks of 2000 rows, and print the number of rows in each chunk as you iterate.

import pandas as pd

data = {
    'product_name': ['Mobile', 'Laptop', 'Shoes', 'Watch', 'Headphones'] * 2000,
    'price': [15000, 45000, 2000, 3000, 1500] * 2000,
    'category': ['Electronics', 'Electronics', 'Fashion', 'Fashion', 'Electronics'] * 2000
}

df = pd.DataFrame(data)
df.to_csv('flipkart.csv')

df = pd.read_csv('flipkart.csv')

df.to_excel('flipkart_products.xlsx', index=False)
from openpyxl import load_workbook
import pandas as pd

wb = load_workbook('flipkart_products.xlsx', read_only=True)
ws = wb.active

rows = ws.iter_rows(values_only=True)

headers = next(rows)

chunk_size = 2000
chunk = []
chunk_number = 0

for row in rows:
    chunk.append(row)

    if len(chunk) == chunk_size:
        chunk_number += 1
        df_chunk = pd.DataFrame(chunk, columns=headers)

        print("Chunk", chunk_number, "Rows:", len(df_chunk))
        chunk = []

if chunk:
    chunk_number += 1
    df_chunk = pd.DataFrame(chunk, columns=headers)
    print("Chunk", chunk_number, "Rows:", len(df_chunk))

wb.close()
```
