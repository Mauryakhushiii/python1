```python
#Install Matplotlib in your Python environment and create a simple line plot showing the number of daily steps you took over the last 7 days. Save the chart as steps_lineplot.png.

import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
steps = [3000, 4500, 4000, 6000, 5500, 8000, 7500]

plt.plot(days, steps, marker='o')

plt.title('Daily Steps for 7 Days')
plt.xlabel('Days')
plt.ylabel('Number of Steps')

plt.grid(True)

plt.savefig('steps_lineplot.png')
plt.show()
```

```python
#Build a bar chart that shows the number of orders you or your friends placed on Swiggy, Zomato, and Domino’s in the last month. Use different colors for each bar and add a legend.

apps = ['Swiggy', 'Zomato', "Domino's"]
orders = [12, 8, 5]

colors = ['orange', 'red', 'blue']

plt.bar(apps, orders, color=colors, label='Number of Orders')

plt.xlabel('Food Delivery Apps')
plt.ylabel('Number of Orders')
plt.title('Orders Placed in Last Month')

plt.legend()
plt.show()
```

```python
#Create a histogram of the durations (in minutes) of your last 20 Spotify listening sessions using Matplotlib. Set the number of bins to 5 and customize the color of the bars.
import matplotlib.pyplot as plt

# Listening duration in minutes (20 sessions)
durations = [15, 20, 25, 30, 35,
             40, 45, 50, 20, 25,
             30, 35, 40, 45, 50,
             55, 60, 65, 70, 75]

plt.hist(durations, bins=5, color='skyblue', edgecolor='black')

plt.xlabel('Listening Duration (Minutes)')
plt.ylabel('Frequency')
plt.title('Spotify Listening Session Durations')

plt.show()
```

```python
#Customize a Matplotlib figure by creating a plot with two subplots: one line plot showing your daily Instagram screen time for a week, and one bar chart showing the number of posts you liked each day. Add appropriate titles, axis labels, and save the figure as social_media_usage.png.<br><br><em><strong>Hint:</strong> Use plt.subplots() to create multiple axes in one figure.</em>
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

# Daily screen time in hours
screen_time = [2, 3, 2.5, 4, 3.5, 5, 4]

# Number of liked posts
liked_posts = [20, 25, 18, 30, 28, 40, 35]

# Create two subplots
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# First subplot: Line plot
ax[0].plot(days, screen_time, marker='o', color='blue')
ax[0].set_title('Daily Instagram Screen Time')
ax[0].set_xlabel('Days')
ax[0].set_ylabel('Screen Time (Hours)')

# Second subplot: Bar chart
ax[1].bar(days, liked_posts, color='orange')
ax[1].set_title('Daily Instagram Liked Posts')
ax[1].set_xlabel('Days')
ax[1].set_ylabel('Number of Likes')

# Adjust layout
plt.tight_layout()

# Save figure
plt.savefig('social_media_usage.png')

# Display figure
plt.show()
```
