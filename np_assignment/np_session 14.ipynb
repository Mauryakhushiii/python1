```python
#Download a small dataset of IPL cricket matches (CSV or Excel), load it into a pandas DataFrame, and perform univariate analysis by printing the summary statistics (mean, median, min, max, std) for the 'total_runs' column.
import pandas as pd

df = pd.read_csv(r'C:/Users/DELL/Downloads/ipl_small_sample.csv')


print(df.head())

print("Mean:", df['total_runs'].mean())
print("Median:", df['total_runs'].median())
print("Min:", df['total_runs'].min())
print("Max:", df['total_runs'].max())
print("Std:", df['total_runs'].std())
```

```python
#
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(r'C:/Users/DELL/Downloads/flipkart_reviews_sample.csv')

# Count reviews for each rating
rating_count = df['rating'].value_counts().sort_index()

# Create bar plot
sns.barplot(x=rating_count.index, y=rating_count.values)

# Add labels and title
plt.xlabel('Rating (Stars)')
plt.ylabel('Number of Reviews')
plt.title('Flipkart Product Reviews by Rating')

# Display plot
plt.show()
```

```python
#Take a dataset of Zomato restaurant listings (with 'average_cost_for_two' and 'user_rating' columns) and create a scatterplot to visualize the relationship between cost and user rating.<br><br><em><strong>Hint:</strong> Use seaborn's scatterplot() function and label the axes clearly.</em>
df = pd.read_csv(r'C:/Users/DELL/Downloads/zomato_restaurants_sample.csv')
sns.scatterplot(
    data=df,
    x='average_cost_for_two',
    y='user_rating'
)

plt.xlabel('Average Cost for Two')
plt.ylabel('User Rating')
plt.title('Zomato Cost vs User Rating')

plt.show()
```

```python
#For a BookMyShow movie dataset (with 'genre' and 'box_office_collection' columns), use groupby analysis to find the average box office collection for each genre and display the results as a sorted table.
import pandas as pd

df = pd.read_csv(r'C:\Users\DELL\Downloads\bookmyshow_movies_sample.csv')

genre_avg = df.groupby('genre')['box_office_collection'].mean()

genre_avg = genre_avg.sort_values(ascending=False)

print(genre_avg)
```
