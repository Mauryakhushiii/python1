```python
#Use plt.subplots() to create a 2x2 grid of subplots and plot four different types of charts (line, bar, scatter, and pie) using any sample data of your choice.
import matplotlib.pyplot as plt

# Create 2x2 subplots
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# Sample data
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
steps = [3000, 4500, 4000, 6000, 7000]

apps = ['Swiggy', 'Zomato', 'Dominos']
orders = [12, 8, 5]

ratings = [3.5, 4.0, 4.5, 3.8, 4.2]
prices = [200, 400, 600, 300, 500]

subjects = ['Python', 'SQL', 'Pandas', 'NumPy']
marks = [80, 70, 90, 60]

# 1. Line Plot
ax[0, 0].plot(days, steps, marker='o', color='blue')
ax[0, 0].set_title('Daily Steps')
ax[0, 0].set_xlabel('Days')
ax[0, 0].set_ylabel('Steps')

# 2. Bar Chart
ax[0, 1].bar(apps, orders, color=['orange', 'red', 'blue'])
ax[0, 1].set_title('Food Orders')
ax[0, 1].set_xlabel('Apps')
ax[0, 1].set_ylabel('Orders')

# 3. Scatter Plot
ax[1, 0].scatter(ratings, prices, color='green')
ax[1, 0].set_title('Ratings vs Price')
ax[1, 0].set_xlabel('Ratings')
ax[1, 0].set_ylabel('Price (₹)')

# 4. Pie Chart
ax[1, 1].pie(marks, labels=subjects, autopct='%1.1f%%')
ax[1, 1].set_title('Subject Marks')

# Adjust layout
plt.tight_layout()

# Save figure
plt.savefig('four_charts.png')

# Display figure
plt.show()
```

```python
#Plot a comparison of average delivery times for Zomato, Swiggy, and Domino's using a bar chart, and style each bar with a different color, linestyle, and linewidth using Matplotlib plot styling options.
import matplotlib.pyplot as plt

# Sample data
apps = ['Zomato', 'Swiggy', "Domino's"]
delivery_time = [30, 25, 40]

colors = ['red', 'orange', 'blue']
linestyles = ['solid', 'dashed', 'dotted']
linewidths = [2, 3, 4]

# Create bar chart
bars = plt.bar(apps, delivery_time, color=colors)

# Customize each bar
for bar, style, width in zip(bars, linestyles, linewidths):
    bar.set_linestyle(style)
    bar.set_linewidth(width)

plt.title('Average Delivery Time Comparison')
plt.xlabel('Food Delivery Apps')
plt.ylabel('Delivery Time (Minutes)')

plt.show()
```

```python
#Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts (right y-axis) for five influencers. Use different colors and markers for each axis.
import matplotlib.pyplot as plt

# Sample data
influencers = ['A', 'B', 'C', 'D', 'E']

followers = [10000, 25000, 18000, 40000, 30000]
daily_posts = [2, 3, 1, 4, 2.5]

# Create figure and first axis
fig, ax1 = plt.subplots(figsize=(10, 5))

# Left Y-axis: Followers
ax1.plot(influencers, followers,
         color='blue', marker='o',
         linewidth=2, label='Followers')

ax1.set_xlabel('Influencers')
ax1.set_ylabel('Instagram Followers', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

# Create second Y-axis
ax2 = ax1.twinx()

# Right Y-axis: Daily Posts
ax2.plot(influencers, daily_posts,
         color='red', marker='s',
         linewidth=2, linestyle='--',
         label='Daily Posts')

ax2.set_ylabel('Average Daily Posts', color='red')
ax2.tick_params(axis='y', labelcolor='red')

# Title
plt.title('Instagram Followers vs Daily Posts')

plt.show()
```

```python
#Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings (numeric) using a scatter plot. Add annotations to display the movie names above each point.
import matplotlib.pyplot as plt

# Sample data
movies = ['Jawan', 'Pathaan', 'Animal', 'Stree 2', 'Dunki']

# IMDB ratings
ratings = [6.9, 5.8, 6.1, 7.0, 6.5]

# Tickets sold (in lakhs)
tickets = [120, 110, 95, 105, 80]

# Create scatter plot
plt.scatter(ratings, tickets, color='blue', marker='o', s=100)

# Add movie name annotations
for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (ratings[i], tickets[i]),
        xytext=(5, 8),
        textcoords='offset points'
    )

# Add labels and title
plt.xlabel('IMDb Ratings')
plt.ylabel('Tickets Sold (Lakhs)')
plt.title('Bollywood Movies: Tickets Sold vs IMDb Ratings')

plt.grid(True)
plt.show()
```

```python
#Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place the label at the correct data point.</em>
import matplotlib.pyplot as plt

# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May',
          'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

sales = [20, 25, 30, 28, 35, 40, 38, 45, 50, 90, 60, 55]

# Create figure and axis
fig, ax = plt.subplots(figsize=(10, 5))

# Line plot
ax.plot(months, sales, marker='o', color='blue')

# Find highest sales point
max_index = sales.index(max(sales))

# Add annotation
ax.annotate(
    'Big Billion Days',
    xy=(months[max_index], sales[max_index]),
    xytext=(months[max_index], sales[max_index] + 15),
    arrowprops=dict(facecolor='red', arrowstyle='->'),
    fontsize=12,
    color='red'
)

# Labels and title
ax.set_xlabel('Months')
ax.set_ylabel('Monthly Sales (₹ Crores)')
ax.set_title('Flipkart Monthly Sales')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
