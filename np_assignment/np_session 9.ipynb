```python
#Given a list of delivery timestamps as strings (e.g., ['2024-06-01 14:30', '2024-06-02 09:15', '2024-06-03 20:45']), use pandas and pd.to_datetime() to convert them into datetime objects and print the result.
import pandas as pd

timestamps = [
    '2024-06-01 14:30',
    '2024-06-02 09:15',
    '2024-06-03 20:45'
]

df = pd.DataFrame({'delivery_time': timestamps})

df['delivery_time'] = pd.to_datetime(df['delivery_time'])

print(df)
print(df.dtypes)
```

```python
#Load a CSV file containing order dates from a Flipkart-style order history (column: 'order_date', format: 'YYYY-MM-DD HH:MM:SS'). Extract the year, month, and weekday for each order and add them as new columns in the DataFrame.

data = {
    'order_id': [101, 102, 103, 104, 105],
    'order_date': [
        '2024-06-01 14:30:00',
        '2024-06-02 09:15:00',
        '2024-06-03 20:45:00',
        '2024-07-05 11:30:00',
        '2024-08-10 16:20:00'
    ]
}

df = pd.DataFrame(data)

df.to_csv('flipkart_orders.csv', index=False)


df = pd.read_csv('flipkart_orders.csv')

print(df.head())
df['order_date'] = pd.to_datetime(df['order_date'])

df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['weekday'] = df['order_date'].dt.day_name()

print(df)
```

```python
#Set the 'order_date' column as the index of your DataFrame and use resampling to calculate the total number of orders placed each week.<br><br><em><strong>Hint:</strong> Use df.resample('W').size() after setting the datetime index.</em>

df = df.set_index('order_date')
weekly_orders = df.resample('W').size()

print(weekly_orders)
```

```python
#Suppose you have a DataFrame of Instagram posts with a 'posted_at' column in UTC. Convert these timestamps to 'Asia/Kolkata' timezone and display the first 5 converted times.

data = {
    'post_id': [101, 102, 103, 104, 105, 106],
    'posted_at': [
        '2024-06-01 08:30:00',
        '2024-06-02 04:15:00',
        '2024-06-03 14:45:00',
        '2024-06-04 10:00:00',
        '2024-06-05 16:20:00',
        '2024-06-06 12:00:00'
    ]
}

df = pd.DataFrame(data)
df['posted_at'] = pd.to_datetime(df['posted_at'], utc=True)
df['posted_at'] = df['posted_at'].dt.tz_convert('Asia/Kolkata')

print(df['posted_at'].head(5))
```

```python
#Create a new feature called 'is_weekend' in your DataFrame that marks True if an order was placed on Saturday or Sunday, and False otherwise.<br><br><em><strong>Constraint:</strong> Do not use any external libraries except pandas and numpy.</em>
import numpy as np

data = {
    'order_id': [101, 102, 103, 104, 105],
    'order_date': [
        '2024-06-01',
        '2024-06-02',
        '2024-06-03',
        '2024-06-07',
        '2024-06-08'
    ]
}

df = pd.DataFrame(data)

df['order_date'] = pd.to_datetime(df['order_date'])
df['is_weekend'] = df['order_date'].dt.day_of_week.isin([5, 6])

print(df)
```
